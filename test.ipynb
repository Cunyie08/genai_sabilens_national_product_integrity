```python
from ultralytics import YOLO

# Load the trained model (replace with the correct path to your trained model)
model = YOLO("runs/detect/train8/weights/best.pt")  # Replace with the correct path to your trained model

# Test image path (replace with the path to your test image)
image_path = "images (3).jpeg"  # Replace with the correct test image path

# Run inference on the test image
results = model(image_path)  # This will return a list of detection results

# Access the first result (for this example, just the first image)
detections = results[0]

# If you want to view the image with predictions and bounding boxes
detections.plot()  # This will plot the image with the bounding boxes

# Optionally, you can save the output to a folder
detections.save()  # Saves the image with bounding boxes in the 'runs/detect' folder

# Print detailed detection results (including labels and coordinates)
print("Detection Results:")
for result in detections.boxes:
    label = result.cls  # The class label index (this is an integer)
    confidence = result.conf.item()  # Convert tensor to float using .item()
    coords = result.xyxy  # Bounding box coordinates (x1, y1, x2, y2)

    # Get the name of the class label based on the class index
    class_name = detections.names[label.item()]  # Convert the index to the corresponding class name

    print(f"Label: {class_name} | Confidence: {confidence:.4f} | Coordinates: {coords.tolist()}")

    # Here, you can specifically check for product-related details
    if class_name == "product name":
        print("Product Name Detected!")
    elif class_name == "product variant":
        print("Product Variant Detected!")
    elif class_name == "NAFDAC number":
        print("NAFDAC Number Detected!")
    elif class_name == "Expiry date":
        print("Expiry Date Detected!")
    elif class_name == "Barcodes":
        print("Barcode Detected!")
```

```python
from ultralytics import YOLO
import cv2
import pytesseract  # Make sure pytesseract is installed

# Load the trained model
model = YOLO("runs/detect/train8/weights/best.pt")  # Path to your trained model

# Test image path (replace with the correct test image path)
image_path = "images (4).jpeg"  # Replace with the correct test image path

# Run inference on the test image
results = model(image_path)

# Access the first result (for this example, just the first image)
detections = results[0]

# If you want to view the image with predictions and bounding boxes
detections.plot()

# Optionally, save the output to a folder
detections.save()

# Extract the detected bounding boxes
for result in detections.boxes:
    label = result.cls  # The class label index (this is an integer)
    confidence = result.conf.item()  # Convert tensor to float using .item()
    coords = result.xyxy  # Bounding box coordinates (x1, y1, x2, y2)

    # Get the name of the class label based on the class index
    class_name = detections.names[label.item()]

    print(f"Label: {class_name} | Confidence: {confidence:.4f} | Coordinates: {coords.tolist()}")

    # Check if the detected label is "product name"
    if class_name == "product name":
        # Crop the image based on the bounding box coordinates
        x1, y1, x2, y2 = map(int, coords[0])
        cropped_image = cv2.imread(image_path)
        cropped_image = cropped_image[y1:y2, x1:x2]  # Crop the region

        # Use Tesseract OCR to extract text from the cropped image
        product_name = pytesseract.image_to_string(cropped_image)

        # Print the detected product name
        print(f"Detected Product Name: {product_name.strip()}")
```

```python
from ultralytics import YOLO
import pytesseract
import cv2
from PIL import Image

# Load the trained YOLO model (make sure the path to your model is correct)
model = YOLO("runs/detect/train8/weights/best.pt")  # Replace with the correct path to your trained model

# Test image path
image_path = "images (4).jpeg"  # Replace with the correct test image path

# Run inference on the test image
results = model(image_path)

# Access the first result (assuming the image only has one set of detections)
boxes = results[0].boxes.xyxy  # Get the bounding boxes of detected objects

# Loop through the boxes to check for product name detection
for box in boxes:
    label = int(results[0].boxes.cls[0])  # Get the class label
    confidence = results[0].boxes.conf[0]  # Confidence score for the detection

    # If it's the "product name" label, apply OCR to extract text
    if results[0].names[label] == "product name":
        x1, y1, x2, y2 = box
        img = cv2.imread(image_path)
        cropped_img = img[int(y1):int(y2), int(x1):int(x2)]  # Crop the detected region

        # Convert to grayscale for better OCR performance
        gray_cropped_img = cv2.cvtColor(cropped_img, cv2.COLOR_BGR2GRAY)

        # Use Tesseract OCR to extract text from the cropped region
        text = pytesseract.image_to_string(gray_cropped_img)

        # Print the detected text (Product Name)
        print(f"Detected Product Name: {text.strip()}")

    # Print other detected information
    print(f"Label: {results[0].names[label]} | Confidence: {confidence:.4f} | Coordinates: {box.tolist()}")

# Optionally, you can visualize the results
results[0].plot()  # Show image with bounding boxes and labels
```
